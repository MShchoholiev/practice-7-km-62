$$\log_a x = \frac{\ln x}{\ln a}.$$

   - якщо введене значення неможливо перетворити на число:

     `Помилка введення: потрібно ввести числа.`

     `Помилка області визначення: x та a мають бути додатними.`

     `Помилка ділення на нуль: основа логарифма не може дорівнювати 1.`
